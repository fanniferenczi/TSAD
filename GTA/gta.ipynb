```python
import sys, torch, torch.nn as nn, types, math, numpy as np
sys.path.insert(0, './GTA')

from models.gta import GTA
from fvcore.nn import FlopCountAnalysis

# ── Patch ProbAttention.forward to fix np.astype on traced Tensor ─────────
# Original lines:
#   U = self.factor * np.ceil(np.log(S)).astype('int').item()
#   u = self.factor * np.ceil(np.log(L)).astype('int').item()
# Fix: convert S and L to Python ints before passing to numpy

def patched_prob_attention_forward(self, queries, keys, values, attn_mask):
    B, L, H, D = queries.shape
    _, S, _, _ = keys.shape

    # ── FIX 1: convert to Python int for numpy compatibility during tracing ──
    S_int = int(S)
    L_int = int(L)

    queries = queries.view(B, H, L_int, -1)
    keys    = keys.view(B, H, S_int, -1)
    values  = values.view(B, H, S_int, -1)

    U = self.factor * np.ceil(np.log(S_int)).astype('int').item()
    u = self.factor * np.ceil(np.log(L_int)).astype('int').item()

    # clamp to valid ranges
    U = min(U, S_int)
    u = min(u, L_int)

    # ── FIX 2: patch _prob_QK inline to fix squeeze() with batch=1 ──────────
    # Original: Q_K_sample.squeeze() collapses batch dim when B=1
    # Fix: use squeeze(-2) to only remove the matmul dim
    def safe_prob_QK(Q, K, sample_k, n_top):
        B_, H_, L_, E_ = K.shape
        _, _, S_, _ = Q.shape
        K_expand = K.unsqueeze(-3).expand(B_, H_, S_, L_, E_)
        indx_sample = torch.randint(L_, (S_, sample_k))
        K_sample = K_expand[:, :, torch.arange(S_).unsqueeze(1), indx_sample, :]
        Q_K_sample = torch.matmul(
            Q.unsqueeze(-2), K_sample.transpose(-2, -1)
        ).squeeze(-2)  # ← only squeeze the matmul dim, not batch dim

        M = Q_K_sample.max(-1)[0] - torch.div(Q_K_sample.sum(-1), L_)
        M_top = M.topk(min(n_top, M.shape[-1]), sorted=False)[1]

        Q_reduce = Q[torch.arange(B_)[:, None, None],
                     torch.arange(H_)[None, :, None],
                     M_top, :]
        Q_K = torch.matmul(Q_reduce, K.transpose(-2, -1))
        return Q_K, M_top

    scores_top, index = safe_prob_QK(queries, keys, u, U)

    scale = self.scale or 1./math.sqrt(D // H)
    if scale is not None:
        scores_top = scores_top * scale

    context = self._get_initial_context(values, L_int)
    context = self._update_context(context, values, scores_top, index, L_int, attn_mask)

    return context.contiguous()

# ── Model setup ───────────────────────────────────────────────────────────
num_nodes  = 25
seq_len    = 60
label_len  = 30
pred_len   = 1
num_levels = 3
factor     = 5
d_model    = 128
n_heads    = 8
e_layers   = 3
d_layers   = 2
d_ff       = 128
dropout    = 0.0
attn       = 'prob'
embed      = 'fixed'
data       = 'SMAP'
activation = 'gelu'
device     = torch.device('cpu')

model = GTA(
    num_nodes, seq_len, label_len, pred_len, num_levels,
    factor, d_model, n_heads, e_layers, d_layers, d_ff,
    dropout, attn, embed, data, activation, device
).double().cpu().eval()

# Apply patch to every ProbAttention module in the model
for module in model.modules():
    if type(module).__name__ == 'ProbAttention':
        module.forward = types.MethodType(patched_prob_attention_forward, module)

# ── Wrapper ───────────────────────────────────────────────────────────────
class GTAWrapper(nn.Module):
    def __init__(self, model):
        super().__init__()
        self.model = model

    def forward(self, x, y, x_mark, y_mark):
        out = self.model(x, y, x_mark, y_mark)
        return out[0] if isinstance(out, tuple) else out

wrapper = GTAWrapper(model)

batch  = 1
x      = torch.randn(batch, seq_len, num_nodes).double()
y      = torch.randn(batch, label_len + pred_len, num_nodes).double()
x_mark = torch.randn(batch, seq_len, 6).double()
y_mark = torch.randn(batch, label_len + pred_len, 6).double()

# ── fvcore analysis ───────────────────────────────────────────────────────
fa = FlopCountAnalysis(wrapper, (x, y, x_mark, y_mark))
fa.unsupported_ops_warnings(False)
fa.uncalled_modules_warnings(False)
_ = fa.total()

by_mod      = fa.by_module()
unsupported = fa.unsupported_ops()

# ── Coverage report ───────────────────────────────────────────────────────
print("=" * 90)
print(f"  fvcore coverage  |  num_nodes={num_nodes}  seq_len={seq_len}  d_model={d_model}")
print(f"  COUNTED              = fvcore has a handler and recorded > 0 MACs")
print(f"  NOT COUNTED (0)      = module executed but fvcore recorded 0 MACs")
print(f"  NOT COUNTED (unsupported op) = raw tensor op with no fvcore handler")
print("=" * 90)

print("\n── Module-level coverage ───────────────────────────────────────────────────────────")
print(f"  {'Module':<65} {'Type':<25} {'MACs':>12}  {'Status'}")
print(f"  {'-'*65} {'-'*25} {'-'*12}  {'-'*20}")

for name, module in wrapper.named_modules():
    if name == '':
        continue
    if len(list(module.children())) > 0:
        continue

    layer_type = type(module).__name__
    macs       = by_mod.get(name, None)

    if macs is None:
        status   = 'NOT COUNTED (not traced)'
        macs_str = '-'
    elif macs > 0:
        status   = 'COUNTED'
        macs_str = f'{macs:,}'
    else:
        status   = 'NOT COUNTED (0)'
        macs_str = '0'

    depth        = name.count('.')
    indent       = '  ' * depth
    display_name = indent + name.split('.')[-1]

    print(f"  {display_name:<65} {layer_type:<25} {macs_str:>12}  {status}")

print("\n── Op-level coverage (raw tensor ops with no fvcore handler) ───────────────────────")
print(f"  {'Operator':<55} {'Occurrences':>12}  {'Status'}")
print(f"  {'-'*55} {'-'*12}  {'-'*15}")
for op, count in sorted(unsupported.items(), key=lambda x: -x[1]):
    print(f"  {op:<55} {count:>12}x  NOT COUNTED")

print("\n── Total ───────────────────────────────────────────────────────────────────────────")
print(f"  MACs recorded by fvcore:         {fa.total():>12,}")
print(f"  Unsupported op types:            {len(unsupported):>12,}")
print(f"  Total unsupported op calls:      {sum(unsupported.values()):>12,}")
print("=" * 90)
```

```python
import sys, torch, torch.nn as nn, types, math, numpy as np
sys.path.insert(0, './GTA')

from models.gta import GTA
from fvcore.nn import FlopCountAnalysis

# ── Patch ProbAttention.forward ────────────────────────────────────────────
def patched_prob_attention_forward(self, queries, keys, values, attn_mask):
    B, L, H, D = queries.shape
    _, S, _, _ = keys.shape

    S_int = int(S)
    L_int = int(L)

    queries = queries.view(B, H, L_int, -1)
    keys    = keys.view(B, H, S_int, -1)
    values  = values.view(B, H, S_int, -1)

    U = self.factor * np.ceil(np.log(S_int)).astype('int').item()
    u = self.factor * np.ceil(np.log(L_int)).astype('int').item()
    U = min(U, S_int)
    u = min(u, L_int)

    def safe_prob_QK(Q, K, sample_k, n_top):
        B_, H_, L_, E_ = K.shape
        _, _, S_, _ = Q.shape
        K_expand = K.unsqueeze(-3).expand(B_, H_, S_, L_, E_)
        indx_sample = torch.randint(L_, (S_, sample_k))
        K_sample = K_expand[:, :, torch.arange(S_).unsqueeze(1), indx_sample, :]
        Q_K_sample = torch.matmul(
            Q.unsqueeze(-2), K_sample.transpose(-2, -1)
        ).squeeze(-2)
        M = Q_K_sample.max(-1)[0] - torch.div(Q_K_sample.sum(-1), L_)
        M_top = M.topk(min(n_top, M.shape[-1]), sorted=False)[1]
        Q_reduce = Q[torch.arange(B_)[:, None, None],
                     torch.arange(H_)[None, :, None],
                     M_top, :]
        Q_K = torch.matmul(Q_reduce, K.transpose(-2, -1))
        return Q_K, M_top

    scores_top, index = safe_prob_QK(queries, keys, u, U)
    scale = self.scale or 1./math.sqrt(D // H)
    if scale is not None:
        scores_top = scores_top * scale
    context = self._get_initial_context(values, L_int)
    context = self._update_context(context, values, scores_top, index, L_int, attn_mask)
    return context.contiguous()

# ── Handler helpers ────────────────────────────────────────────────────────
def get_shape(jit_value):
    return list(jit_value.type().sizes())

def numel_from_jit(jit_value):
    shape = get_shape(jit_value)
    result = 1
    for s in shape:
        result *= s
    return result

def elementwise_handler(inputs, outputs):
    """1 op per output element — add, mul, sub, div, pad, pow_, log, neg."""
    try:
        return numel_from_jit(outputs[0])
    except Exception:
        return 0

def softmax_handler(inputs, outputs):
    """
    Softmax: exp + sum + divide ≈ 2 ops per element.
    Consistent with Anomaly Transformer handler.
    """
    try:
        return 2 * numel_from_jit(outputs[0])
    except Exception:
        return 0

def gelu_handler(inputs, outputs):
    """
    GELU ≈ 5 ops per element (mul, pow, mul, add, tanh).
    Consistent with Anomaly Transformer handler.
    """
    try:
        return 5 * numel_from_jit(outputs[0])
    except Exception:
        return 0

def elu_handler(inputs, outputs):
    """
    ELU: max(0,x) + min(0, α*(exp(x)-1))
    ≈ 3 ops per element (exp, sub, conditional mul).
    Used in distilled conv layers.
    """
    try:
        return 3 * numel_from_jit(outputs[0])
    except Exception:
        return 0

def sum_handler(inputs, outputs):
    """
    Reduction sum: 1 op per input element.
    Used in ProbAttention sparsity measurement.
    """
    try:
        shape = get_shape(inputs[0])
        result = 1
        for s in shape:
            result *= s
        return result
    except Exception:
        return 0

def zero_handler(inputs, outputs):
    """Explicitly return 0 for ops with no arithmetic compute."""
    return 0

# ── Apply all handlers ─────────────────────────────────────────────────────
def apply_handlers(fa):
    # --- Non-trivial compute: add handlers ---
    fa.set_op_handle('aten::add',          elementwise_handler)  # 23x residuals
    fa.set_op_handle('aten::mul',          elementwise_handler)  # 19x scaling
    fa.set_op_handle('aten::pad',          elementwise_handler)  # 10x conv padding
    fa.set_op_handle('aten::softmax',      softmax_handler)      # 10x attention
    fa.set_op_handle('aten::div',          elementwise_handler)  # 6x scaling
    fa.set_op_handle('aten::sub',          elementwise_handler)  # 6x IP conv
    fa.set_op_handle('aten::sum',          sum_handler)          # 6x sparsity measure
    fa.set_op_handle('aten::gelu',         gelu_handler)         # 5x FFN
    fa.set_op_handle('aten::pow_',         elementwise_handler)  # 3x gumbel
    fa.set_op_handle('aten::log',          elementwise_handler)  # 3x gumbel
    fa.set_op_handle('aten::neg',          elementwise_handler)  # 3x gumbel
    fa.set_op_handle('aten::add_',         elementwise_handler)  # 3x in-place residuals
    fa.set_op_handle('aten::elu',          elu_handler)          # 2x distilled conv

    # --- Negligible / no arithmetic: return 0 explicitly ---
    fa.set_op_handle('aten::_weight_norm', zero_handler)   # weight norm recompute
    fa.set_op_handle('aten::expand_as',    zero_handler)   # memory view
    fa.set_op_handle('aten::scatter_add_', zero_handler)   # index scatter
    fa.set_op_handle('aten::masked_fill_', zero_handler)   # boolean mask
    fa.set_op_handle('aten::empty_like',   zero_handler)   # allocation
    fa.set_op_handle('aten::exponential_', zero_handler)   # in Gumbel manual count
    fa.set_op_handle('aten::scatter_',     zero_handler)   # index scatter
    fa.set_op_handle('aten::index_select', zero_handler)   # index gather
    fa.set_op_handle('aten::randint',      zero_handler)   # random index gen
    fa.set_op_handle('aten::topk',         zero_handler)   # sort, negligible
    fa.set_op_handle('aten::clone',        zero_handler)   # memory copy
    fa.set_op_handle('aten::max_pool1d',   zero_handler)   # comparison, negligible
    fa.set_op_handle('aten::triu',         zero_handler)   # mask generation
    return fa

# ── Gumbel-Softmax MACs: manual calculation ────────────────────────────────
# The Gumbel-Softmax graph structure learning (Chen et al., 2022, §IV-A)
# samples M*(M-1) binary connection variables using Gumbel noise.
# Per directed pair (i,j), the operations are:
#   u ~ Uniform(0,1):          1 op  (exponential_ in log space)
#   g = -log(-log(u)):         3 ops (log, neg, log)
#   z = (log_π + g) / τ:       2 ops (add, div)
#   softmax over {0,1}:        2 ops (exp + div per element)
# Total: ~8 ops per directed pair, times M*(M-1) pairs
GUMBEL_OPS_PER_PAIR = 8

def gumbel_softmax_macs(num_nodes):
    """
    MACs for Gumbel-Softmax graph structure learning.
    num_nodes = M (number of sensors).
    """
    n_pairs = num_nodes * (num_nodes - 1)
    return GUMBEL_OPS_PER_PAIR * n_pairs

# ── Main measurement function ──────────────────────────────────────────────
def run_fvcore_gta(num_nodes, seq_len=60, label_len=30, pred_len=1,
                   num_levels=3, factor=5, d_model=128, n_heads=8,
                   e_layers=3, d_layers=2, d_ff=128):

    model = GTA(
        num_nodes, seq_len, label_len, pred_len, num_levels,
        factor, d_model, n_heads, e_layers, d_layers, d_ff,
        0.0, 'prob', 'fixed', 'SMAP', 'gelu', torch.device('cpu')
    ).double().cpu().eval()

    # apply patch to all ProbAttention modules
    for module in model.modules():
        if type(module).__name__ == 'ProbAttention':
            module.forward = types.MethodType(
                patched_prob_attention_forward, module
            )

    class GTAWrapper(nn.Module):
        def __init__(self, model):
            super().__init__()
            self.model = model
        def forward(self, x, y, x_mark, y_mark):
            out = self.model(x, y, x_mark, y_mark)
            return out[0] if isinstance(out, tuple) else out

    wrapper = GTAWrapper(model)
    x      = torch.randn(1, seq_len, num_nodes).double()
    y      = torch.randn(1, label_len + pred_len, num_nodes).double()
    x_mark = torch.randn(1, seq_len, 6).double()
    y_mark = torch.randn(1, label_len + pred_len, 6).double()

    # ── Baseline: fvcore default (Linear + Conv only) ─────────────────────
    fa_base = FlopCountAnalysis(wrapper, (x, y, x_mark, y_mark))
    fa_base.unsupported_ops_warnings(False)
    fa_base.uncalled_modules_warnings(False)
    base_macs   = fa_base.total()
    unsupported = fa_base.unsupported_ops()

    # ── Extended: add handlers for missing ops ─────────────────────────────
    fa_full = FlopCountAnalysis(wrapper, (x, y, x_mark, y_mark))
    fa_full.unsupported_ops_warnings(False)
    fa_full.uncalled_modules_warnings(False)
    fa_full = apply_handlers(fa_full)

    fvcore_macs = fa_full.total()
    by_op_full  = fa_full.by_operator()

    # ── Add Gumbel-Softmax MACs manually ──────────────────────────────────
    # aten::exponential_, aten::log, aten::neg are not reliably traceable
    # because they are in-place ops on the Gumbel noise buffer.
    # They are zeroed in the handler and accounted for analytically here.
    gumbel_macs = gumbel_softmax_macs(num_nodes)
    full_macs   = fvcore_macs + gumbel_macs

    return base_macs, fvcore_macs, gumbel_macs, full_macs, by_op_full, unsupported

# ── Datasets ───────────────────────────────────────────────────────────────
DATASETS = {
    'SMAP':  dict(num_nodes=25),
    'MSL':   dict(num_nodes=55),
    'SMD':   dict(num_nodes=38),
    'SWaT':  dict(num_nodes=51),
    'PSM':   dict(num_nodes=25),
}

default_ops = {
    'conv', 'addmm', 'batch_norm', 'bmm', 'einsum',
    'layer_norm', 'linear', 'matmul', 'mm',
}

for name, hp in DATASETS.items():
    num_nodes = hp['num_nodes']
    base_macs, fvcore_macs, gumbel_macs, full_macs, by_op_full, unsupported = \
        run_fvcore_gta(num_nodes)

    print("=" * 72)
    print(f"  {name}  |  num_nodes={num_nodes}  seq_len=60  d_model=128  e_layers=3")
    print("=" * 72)

    print(f"\n{'Baseline fvcore (Linear + Conv only)':.<52} {base_macs:>12,}")
    print(f"{'Extended fvcore (+ elementwise handlers)':.<52} {fvcore_macs:>12,}")
    print(f"{'Gumbel-Softmax (manual, 8 ops × M*(M-1))':.<52} {gumbel_macs:>12,}")
    print(f"{'Total MACs':.<52} {full_macs:>12,}")

    print("\n── Per-operator breakdown (extended fvcore) ────────────────────────")
    print(f"  {'Operator':<35} {'MACs':>12}  {'Counted by default?'}")
    print(f"  {'-'*35} {'-'*12}  {'-'*25}")
    for op, count in sorted(by_op_full.items(), key=lambda x: -x[1]):
        if count == 0:
            continue
        was_default = 'yes' if op in default_ops else 'no — custom handler'
        print(f"  {op:<35} {count:>12,}  {was_default}")

    print(f"\n  {'gumbel_softmax (manual)':.<35} {gumbel_macs:>12,}  no — analytical")

    print("\n── Previously unsupported ops, now handled ─────────────────────────")
    print(f"  {'Operator':<35} {'Occurrences':>12}  {'MACs now':>12}")
    print(f"  {'-'*35} {'-'*12}  {'-'*12}")
    for op_key, occ in sorted(unsupported.items(), key=lambda x: -x[1]):
        simplified = op_key.replace('aten::', '')
        new_count  = by_op_full.get(simplified, 0)
        print(f"  {op_key:<35} {occ:>12}x  {new_count:>12,}")
    print(f"  {'gumbel_softmax (manual)':.<35} {'—':>12}   {gumbel_macs:>12,}")

    print("\n── Summary ──────────────────────────────────────────────────────────")
    print(f"  Baseline MACs (Linear + Conv):       {base_macs:>12,}")
    print(f"  Added by elementwise handlers:       {fvcore_macs - base_macs:>12,}")
    print(f"  Added by Gumbel-Softmax (manual):    {gumbel_macs:>12,}")
    print(f"  Total MACs:                          {full_macs:>12,}")

    pct_gumbel = gumbel_macs / full_macs * 100 if full_macs > 0 else 0
    pct_elem   = (fvcore_macs - base_macs) / full_macs * 100 if full_macs > 0 else 0
    print(f"  Gumbel-Softmax share:                {pct_gumbel:>11.1f}%")
    print(f"  Elementwise share:                   {pct_elem:>11.1f}%")

    macs_per_sample = full_macs / 60  # seq_len=60
    print(f"\n  GMACs (full window):                 {full_macs/1e9:>12.4f}")
    print(f"  GMACs/sample (÷ seq_len):            {macs_per_sample/1e9:>12.6f}")
    print()
```

## Memory Usage

```python
import sys, torch, torch.nn as nn
sys.path.insert(0, './GTA')

from models.gta import GTA

device = torch.device("cuda:3")  # adjust to your available GPU

# Force CUDA context once before measurements
_ = torch.zeros(1).to(device)
torch.cuda.synchronize()

# ── GTA hyperparameters (paper settings, Chen et al., 2022, §V-B-4) ──────
SEQ_LEN    = 60
LABEL_LEN  = 30
PRED_LEN   = 1
NUM_LEVELS = 3
FACTOR     = 5
D_MODEL    = 128
N_HEADS    = 8
E_LAYERS   = 3
D_LAYERS   = 2
D_FF       = 128
DROPOUT    = 0.05
ATTN       = 'prob'
EMBED      = 'fixed'
ACTIVATION = 'gelu'
BATCH_SIZE = 128  # consistent with cross-model comparability

DATASETS = {
    'SMD':  dict(num_nodes=38),
    'MSL':  dict(num_nodes=55),
    'SMAP': dict(num_nodes=25),
    'SWaT': dict(num_nodes=51),
    'PSM':  dict(num_nodes=25),
}

results = {}

for name, cfg in DATASETS.items():
    num_nodes = cfg['num_nodes']

    # ── Params (CPU, float64 to match .double() runtime) ─────────────────
    model_cpu = GTA(
        num_nodes, SEQ_LEN, LABEL_LEN, PRED_LEN, NUM_LEVELS,
        FACTOR, D_MODEL, N_HEADS, E_LAYERS, D_LAYERS, D_FF,
        DROPOUT, ATTN, EMBED, name, ACTIVATION, torch.device('cpu')
    ).double()
    total_params = sum(p.numel() for p in model_cpu.parameters())
    # float64 = 8 bytes per parameter
    params_mb = total_params * 8 / 1024**2
    del model_cpu

    # ── GPU allocated (weights + buffers on GPU) ──────────────────────────
    torch.cuda.empty_cache()
    torch.cuda.synchronize()
    baseline = torch.cuda.memory_allocated(device)

    model = GTA(
        num_nodes, SEQ_LEN, LABEL_LEN, PRED_LEN, NUM_LEVELS,
        FACTOR, D_MODEL, N_HEADS, E_LAYERS, D_LAYERS, D_FF,
        DROPOUT, ATTN, EMBED, name, ACTIVATION, device
    ).double().to(device)

    torch.cuda.synchronize()
    gpu_allocated = (torch.cuda.memory_allocated(device) - baseline) / 1024**2

    # ── Prepare inputs ────────────────────────────────────────────────────
    # GTA forward: (x, y, x_mark, y_mark)
    # x:      (batch, seq_len, num_nodes)
    # y:      (batch, label_len + pred_len, num_nodes)
    # x_mark: (batch, seq_len, 6)
    # y_mark: (batch, label_len + pred_len, 6)
    x      = torch.randn(BATCH_SIZE, SEQ_LEN, num_nodes,
                         device=device, dtype=torch.float64)
    y      = torch.randn(BATCH_SIZE, LABEL_LEN + PRED_LEN, num_nodes,
                         device=device, dtype=torch.float64)
    x_mark = torch.randn(BATCH_SIZE, SEQ_LEN, 6,
                         device=device, dtype=torch.float64)
    y_mark = torch.randn(BATCH_SIZE, LABEL_LEN + PRED_LEN, 6,
                         device=device, dtype=torch.float64)

    criterion = nn.MSELoss()

    # ── Peak training memory ──────────────────────────────────────────────
    # GTA training loss includes sparsity regularization term:
    # loss = MSE + sum(|logits[:,0]|)  (Chen et al., 2022, §IV-B, Eq. 4)
    model.train()
    optimizer = torch.optim.Adam(model.parameters(), lr=1e-4)

    torch.cuda.empty_cache()
    torch.cuda.synchronize()
    torch.cuda.reset_peak_memory_stats(device)

    model.zero_grad(set_to_none=True)
    output = model(x, y, x_mark, y_mark)
    # GTA returns output directly (not a tuple)
    if isinstance(output, tuple):
        output = output[0]
    # target is the last pred_len steps of y
    target = y[:, -PRED_LEN:, :]
    loss = criterion(output, target) + \
           torch.sum(torch.abs(model.gt_embedding.gc_module.logits[:, 0]))
    loss.backward()
    optimizer.step()

    torch.cuda.synchronize()
    peak_train = torch.cuda.max_memory_allocated(device) / 1024**2

    # ── Peak inference memory ─────────────────────────────────────────────
    model.eval()
    torch.cuda.empty_cache()
    torch.cuda.synchronize()
    torch.cuda.reset_peak_memory_stats(device)

    with torch.no_grad():
        output_inf = model(x, y, x_mark, y_mark)

    torch.cuda.synchronize()
    peak_infer = torch.cuda.max_memory_allocated(device) / 1024**2

    results[name] = dict(
        params=total_params,
        params_mb=params_mb,
        gpu_allocated=gpu_allocated,
        peak_train=peak_train,
        overhead=peak_train - gpu_allocated,
        peak_infer=peak_infer,
    )

    del model, optimizer, x, y, x_mark, y_mark, output, loss
    torch.cuda.empty_cache()

# ── Print table ───────────────────────────────────────────────────────────
print(f"\n{'='*100}")
print(f"MEMORY SUMMARY — GTA (batch_size={BATCH_SIZE}, float64, seq_len={SEQ_LEN})")
print(f"{'='*100}")
print(f"{'Dataset':<8} {'Params':>12} {'Params MB':>10} {'GPU alloc':>10} "
      f"{'Peak train':>12} {'Overhead':>10} {'Peak infer':>12}")
print(f"{'-'*100}")
for name, r in results.items():
    print(f"{name:<8} {r['params']:>12,} {r['params_mb']:>9.2f} MB "
          f"{r['gpu_allocated']:>9.2f} MB {r['peak_train']:>11.2f} MB "
          f"{r['overhead']:>9.2f} MB {r['peak_infer']:>11.2f} MB")
```

## Inference Speed

Measures latency, throughput, and s/iter for each dataset on CPU.
- Latency: time to process one batch (ms)
- Throughput: samples processed per second
- s/iter: latency in seconds

```python
import sys, torch, numpy as np
sys.path.insert(0, './GTA')

from models.gta import GTA

device = torch.device("cuda:1")  # adjust to your available GPU

# ── Force CUDA context initialization ─────────────────────────────────────
_ = torch.zeros(1).to(device)

# ── GTA hyperparameters (paper settings, Chen et al., 2022, §V-B-4) ──────
SEQ_LEN    = 60
LABEL_LEN  = 30
PRED_LEN   = 1
NUM_LEVELS = 3
FACTOR     = 5
D_MODEL    = 128
N_HEADS    = 8
E_LAYERS   = 3
D_LAYERS   = 2
D_FF       = 128
DROPOUT    = 0.05
ATTN       = 'prob'
EMBED      = 'fixed'
ACTIVATION = 'gelu'

BATCH_SIZE = 128
N_RUNS     = 100
N_WARMUP   = 10

DATASETS = {
    'SMD':  dict(num_nodes=38),
    'MSL':  dict(num_nodes=55),
    'SMAP': dict(num_nodes=25),
    'SWaT': dict(num_nodes=51),
    'PSM':  dict(num_nodes=25),
}

results = {}

for name, cfg in DATASETS.items():
    num_nodes = cfg['num_nodes']

    # ── Build model ────────────────────────────────────────────────────────
    model = GTA(
        num_nodes, SEQ_LEN, LABEL_LEN, PRED_LEN, NUM_LEVELS,
        FACTOR, D_MODEL, N_HEADS, E_LAYERS, D_LAYERS, D_FF,
        DROPOUT, ATTN, EMBED, name, ACTIVATION, device
    ).double().to(device)
    model.eval()

    # GTA forward: (x, y, x_mark, y_mark)
    x      = torch.randn(BATCH_SIZE, SEQ_LEN, num_nodes,
                         device=device, dtype=torch.float64)
    y      = torch.randn(BATCH_SIZE, LABEL_LEN + PRED_LEN, num_nodes,
                         device=device, dtype=torch.float64)
    x_mark = torch.randn(BATCH_SIZE, SEQ_LEN, 6,
                         device=device, dtype=torch.float64)
    y_mark = torch.randn(BATCH_SIZE, LABEL_LEN + PRED_LEN, 6,
                         device=device, dtype=torch.float64)

    # ── Warmup runs ────────────────────────────────────────────────────────
    with torch.no_grad():
        for _ in range(N_WARMUP):
            _ = model(x, y, x_mark, y_mark)

    # ── Timed runs using CUDA Events ───────────────────────────────────────
    latencies = []

    with torch.no_grad():
        for _ in range(N_RUNS):
            start = torch.cuda.Event(enable_timing=True)
            end   = torch.cuda.Event(enable_timing=True)

            start.record()
            _ = model(x, y, x_mark, y_mark)
            end.record()

            torch.cuda.synchronize()
            latencies.append(start.elapsed_time(end))  # milliseconds

    latencies  = np.array(latencies)
    mean_ms    = latencies.mean()
    std_ms     = latencies.std()
    throughput = BATCH_SIZE / (mean_ms / 1000)  # samples per second
    s_per_iter = mean_ms / 1000                 # seconds per iteration

    results[name] = {
        'num_nodes':  num_nodes,
        'mean_ms':    mean_ms,
        'std_ms':     std_ms,
        'throughput': throughput,
        's_per_iter': s_per_iter,
    }

    del model, x, y, x_mark, y_mark
    torch.cuda.empty_cache()

# ── Summary table ──────────────────────────────────────────────────────────
print('\n' + '=' * 100)
print(f'INFERENCE SPEED SUMMARY — GTA '
      f'(batch_size={BATCH_SIZE}, seq_len={SEQ_LEN}, float64, N={N_RUNS} runs)')
print('=' * 100)
print(f"{'Dataset':<8} {'num_nodes':<11} {'B':<5} "
      f"{'Latency mean±std (ms)':<28} {'Throughput (samp/s)':<23} {'s/iter'}")
print('-' * 100)
for name, r in results.items():
    print(
        f"{name:<8} {r['num_nodes']:<11} {BATCH_SIZE:<5} "
        f"{r['mean_ms']:.3f} ± {r['std_ms']:<19.3f}"
        f"{r['throughput']:>20,.1f}     {r['s_per_iter']:.4f}"
    )
```
