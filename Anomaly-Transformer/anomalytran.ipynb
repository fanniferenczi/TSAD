```python
import sys, torch, torch.nn as nn, types, math
sys.path.insert(0, './')
import numpy as np

from model.AnomalyTransformer import AnomalyTransformer
from fvcore.nn import FlopCountAnalysis

# ── Patch (same as before) ─────────────────────────────────────────────────
def patched_inner_attention_forward(self, queries, keys, values, sigma, attn_mask):
    B, L, H, E = queries.shape
    _, S, _, D = values.shape
    scale = 1. / math.sqrt(E)
    scores = torch.einsum("blhe,bshe->bhls", queries, keys)
    sigma = sigma.transpose(1, 2)
    window_size = sigma.shape[-1]
    sigma = sigma.unsqueeze(-1).repeat(1, 1, 1, window_size)
    prior = self.distances.unsqueeze(0).unsqueeze(0).repeat(
        sigma.shape[0], sigma.shape[1], 1, 1
    ).to(sigma.device)
    prior = 1.0 / (math.sqrt(2 * math.pi) * sigma) * torch.exp(
        -prior ** 2 / 2 / (sigma ** 2)
    )
    series = self.dropout(torch.softmax(scores * scale, dim=-1))
    V = torch.einsum("bhls,bshd->blhd", series, values)
    return V.contiguous(), series, prior, sigma

class AnomalyTransformerWrapper(nn.Module):
    def __init__(self, model):
        super().__init__()
        self.model = model
    def forward(self, x):
        out = self.model(x)
        return out[0] if isinstance(out, tuple) else out

# ── Handler helpers ────────────────────────────────────────────────────────
def get_shape(jit_value):
    return list(jit_value.type().sizes())

def numel_from_jit(jit_value):
    shape = get_shape(jit_value)
    result = 1
    for s in shape:
        result *= s
    return result

def elementwise_handler(inputs, outputs):
    """1 op per output element — for add, mul, sub, pad."""
    try:
        return numel_from_jit(outputs[0])
    except Exception:
        return 0

def softmax_handler(inputs, outputs):
    """
    Softmax: exp + sum + divide per element ≈ 2 ops per element.
    Same convention as TranAD.
    """
    try:
        return 2 * numel_from_jit(outputs[0])
    except Exception:
        return 0

def gelu_handler(inputs, outputs):
    """
    GELU: x * 0.5 * (1 + tanh(sqrt(2/π) * (x + 0.044715*x³)))
    Approximation: ~5 ops per element (mul, pow, mul, add, tanh).
    """
    try:
        return 5 * numel_from_jit(outputs[0])
    except Exception:
        return 0

def addmm_handler(inputs, outputs):
    """
    addmm = bias add + matrix multiply.
    Covers sigma_projection which fvcore misses at module level
    but does emit as aten::addmm at op level.
    """
    try:
        input_shape  = get_shape(inputs[1])
        weight_shape = get_shape(inputs[2])
        rows      = input_shape[0]
        inner_dim = input_shape[1]
        cols      = weight_shape[1]
        return rows * inner_dim * cols
    except Exception:
        return 0

# ── Gaussian kernel MACs: manual calculation ───────────────────────────────
# The Gaussian prior-association is computed inside the patched forward,
# which the JIT tracer does not follow — so aten::exp, aten::div, aten::pow
# never appear in the unsupported ops list.
# We calculate them analytically and add them to the fvcore total.
#
# Prior computation per layer (Xu et al., ICLR 2022, Eq. 2):
#   prior = 1/(sqrt(2π)*σ) * exp(-d²/2σ²)
# where d ∈ R^(B×H×L×L), σ ∈ R^(B×H×L×L)
#
# Operations over the B×H×L×L matrix:
#   pow (d²):          B*H*L*L
#   mul (d²/2):        B*H*L*L
#   div (÷σ²):         B*H*L*L
#   neg (-...):        B*H*L*L
#   exp:               B*H*L*L
#   mul (÷σ):          B*H*L*L
#   div (÷sqrt(2π)):   B*H*L*L
#   row-sum (rescale): B*H*L*L
# Total: 8 ops per element in the B×H×L×L matrix
GAUSSIAN_OPS_PER_ELEMENT = 8

def gaussian_kernel_macs(batch, n_heads, seq_len, n_layers):
    """
    MACs for Gaussian prior-association across all layers.
    batch=1 for per-sample MAC counting (consistent with fvcore convention).
    """
    return GAUSSIAN_OPS_PER_ELEMENT * batch * n_heads * seq_len * seq_len * n_layers

# ── Main measurement function ──────────────────────────────────────────────
def run_fvcore(enc_in, seq_len=100, d_model=512, n_heads=8, e_layers=3):

    model = AnomalyTransformer(
        win_size=seq_len, enc_in=enc_in, c_out=enc_in,
        e_layers=e_layers, d_model=d_model, n_heads=n_heads
    ).float().cpu().eval()

    for module in model.modules():
        if type(module).__name__ == 'AnomalyAttention':
            module.forward = types.MethodType(
                patched_inner_attention_forward, module
            )

    wrapper = AnomalyTransformerWrapper(model)
    dummy   = torch.randn(1, seq_len, enc_in)  # batch=1

    # ── Baseline: fvcore default (Linear + Conv only) ─────────────────────
    fa_base = FlopCountAnalysis(wrapper, dummy)
    fa_base.unsupported_ops_warnings(False)
    fa_base.uncalled_modules_warnings(False)
    base_macs   = fa_base.total()
    unsupported = fa_base.unsupported_ops()

    # ── Extended: add handlers for all missing ops ─────────────────────────
    fa_full = FlopCountAnalysis(wrapper, dummy)
    fa_full.unsupported_ops_warnings(False)
    fa_full.uncalled_modules_warnings(False)

    # From coverage report — all 5 unsupported ops
    fa_full.set_op_handle('aten::add',     elementwise_handler)  # 7x: residuals + pos embed
    fa_full.set_op_handle('aten::mul',     elementwise_handler)  # 3x: attention scaling
    fa_full.set_op_handle('aten::softmax', softmax_handler)      # 3x: series-association
    fa_full.set_op_handle('aten::gelu',    gelu_handler)         # 3x: feed-forward activation
    fa_full.set_op_handle('aten::pad',     elementwise_handler)  # 1x: Conv1d padding
    # sigma_projection shows 0 at module level but emits aten::addmm
    fa_full.set_op_handle('aten::addmm',   addmm_handler)

    fvcore_macs = fa_full.total()
    by_op_full  = fa_full.by_operator()

    # ── Add Gaussian kernel MACs manually ─────────────────────────────────
    gaussian_macs = gaussian_kernel_macs(
        batch=1, n_heads=n_heads, seq_len=seq_len, n_layers=e_layers
    )
    full_macs = fvcore_macs + gaussian_macs

    return base_macs, fvcore_macs, gaussian_macs, full_macs, by_op_full, unsupported

# ── Datasets ───────────────────────────────────────────────────────────────
DATASETS = {
    'SMD':   dict(enc_in=38),
    'MSL':   dict(enc_in=55),
    'SMAP':  dict(enc_in=25),
    'SWaT':  dict(enc_in=51),
    'PSM':   dict(enc_in=25),
}

SEQ_LEN  = 100
D_MODEL  = 512
N_HEADS  = 8
E_LAYERS = 3

#default_ops = {'matmul', 'bmm', 'linear', 'addmm', 'mm', 'conv1d', 'conv2d'}
default_ops = {
    'conv',        # aten::_convolution
    'adaptive_avg_pool2d',
    'addmm',
    'batch_norm',
    'bmm',
    'einsum',
    'grid_sampler',
    'group_norm',
    'instance_norm',
    'layer_norm',
    'linear',
    'matmul',
    'mm',
    'upsample_bilinear2d',
    'upsample_nearest2d',
}

for name, hp in DATASETS.items():
    enc_in = hp['enc_in']
    base_macs, fvcore_macs, gaussian_macs, full_macs, by_op_full, unsupported = \
        run_fvcore(enc_in, SEQ_LEN, D_MODEL, N_HEADS, E_LAYERS)

    print("=" * 72)
    print(f"  {name}  |  enc_in={enc_in}  seq_len={SEQ_LEN}  "
          f"d_model={D_MODEL}  e_layers={E_LAYERS}")
    print("=" * 72)

    print(f"\n{'Baseline fvcore (Linear + Conv only)':.<52} {base_macs:>12,}")
    print(f"{'Extended fvcore (+ elementwise handlers)':.<52} {fvcore_macs:>12,}")
    print(f"{'Gaussian kernel (manual, 8 ops × B×H×L×L)':.<52} {gaussian_macs:>12,}")
    print(f"{'Total MACs':.<52} {full_macs:>12,}")

    print("\n── Per-operator breakdown (extended fvcore) ────────────────────────")
    print(f"  {'Operator':<35} {'MACs':>12}  {'Counted by default?'}")
    print(f"  {'-'*35} {'-'*12}  {'-'*25}")
    for op, count in sorted(by_op_full.items(), key=lambda x: -x[1]):
        if count == 0:
            continue
        was_default = 'yes' if op in default_ops else 'no — custom handler'
        print(f"  {op:<35} {count:>12,}  {was_default}")

    print(f"\n  {'gaussian_kernel (manual)':.<35} {gaussian_macs:>12,}  "
          f"no — analytical")

    print("\n── Previously unsupported ops, now handled ─────────────────────────")
    print(f"  {'Operator':<35} {'Occurrences':>12}  {'MACs now':>12}")
    print(f"  {'-'*35} {'-'*12}  {'-'*12}")
    for op_full_key, occ in sorted(unsupported.items(), key=lambda x: -x[1]):
        simplified = op_full_key.replace('aten::', '')
        new_count  = by_op_full.get(simplified, 0)
        print(f"  {op_full_key:<35} {occ:>12}x  {new_count:>12,}")
    print(f"  {'gaussian_kernel (manual)':.<35} {'—':>12}   {gaussian_macs:>12,}")

    print("\n── Summary ──────────────────────────────────────────────────────────")
    print(f"  Baseline MACs (Linear + Conv):       {base_macs:>12,}")
    print(f"  Added by elementwise handlers:       {fvcore_macs - base_macs:>12,}")
    print(f"  Added by Gaussian kernel (manual):   {gaussian_macs:>12,}")
    print(f"  Total MACs:                          {full_macs:>12,}")

    pct_gaussian = gaussian_macs / full_macs * 100 if full_macs > 0 else 0
    pct_elem     = (fvcore_macs - base_macs) / full_macs * 100 if full_macs > 0 else 0
    print(f"  Gaussian kernel share:               {pct_gaussian:>11.1f}%")
    print(f"  Elementwise share:                   {pct_elem:>11.1f}%")

    # ── Streaming normalization ───────────────────────────────────────────
    # MACs per single time step for IoT edge deployment context
    # Consistent with TimesNet and TranAD normalization in this thesis
    macs_per_sample = full_macs / SEQ_LEN
    print(f"\n  GMACs (full window):                 {full_macs/1e9:>12.4f}")
    print(f"  GMACs/sample (÷ seq_len):            {macs_per_sample/1e9:>12.6f}")
    print()


    
```

## Memory Usage

```python
import torch
import torch.nn as nn
from model.AnomalyTransformer import AnomalyTransformer
device = torch.device("cuda:0")

# Force CUDA context once before measurements
_ = torch.zeros(1).to(device)
torch.cuda.synchronize()

DATASETS = {
    "SMD":   dict(enc_in=38),
    "MSL":   dict(enc_in=55),
    "SMAP":  dict(enc_in=25),
    "SWaT":  dict(enc_in=51),
    "PSM":   dict(enc_in=25),
}

results = {}

for name, cfg in DATASETS.items():
    enc_in = cfg["enc_in"]

    # ── Params ───────────────────────────────────────────────────────────
    model_cpu = AnomalyTransformer(win_size=100, enc_in=enc_in, c_out=enc_in, e_layers=3)
    total_params = sum(p.numel() for p in model_cpu.parameters())
    params_mb = total_params * 4 / 1024**2
    del model_cpu

    # ── GPU allocated (weights + buffers) ────────────────────────────────
    torch.cuda.empty_cache()
    torch.cuda.synchronize()
    baseline = torch.cuda.memory_allocated(device)
    model = AnomalyTransformer(win_size=100, enc_in=enc_in, c_out=enc_in, e_layers=3).to(device)
    gpu_allocated = (torch.cuda.memory_allocated(device) - baseline) / 1024**2

    # ── Peak training (full minimax step) ────────────────────────────────
    model.train()
    optimizer = torch.optim.Adam(model.parameters(), lr=1e-4)
    x = torch.randn(128, 100, enc_in, device=device)
    criterion = nn.MSELoss()

    torch.cuda.empty_cache()
    torch.cuda.reset_peak_memory_stats(device)

    model.zero_grad(set_to_none=True)
    output, series, prior, sigma = model(x)
    loss = criterion(output, x)
    loss.backward()

    model.zero_grad(set_to_none=True)
    output2, series2, prior2, sigma2 = model(x)
    loss2 = criterion(output2, x)
    loss2.backward()

    optimizer.step()
    torch.cuda.synchronize()
    peak_train = torch.cuda.max_memory_allocated(device) / 1024**2

    # ── Peak inference ───────────────────────────────────────────────────
    model.eval()
    torch.cuda.empty_cache()
    torch.cuda.synchronize()
    torch.cuda.reset_peak_memory_stats(device)

    with torch.no_grad():
        _, _, _, _ = model(x)

    torch.cuda.synchronize()
    peak_infer = torch.cuda.max_memory_allocated(device) / 1024**2

    results[name] = dict(
        params=total_params,
        params_mb=params_mb,
        gpu_allocated=gpu_allocated,
        peak_train=peak_train,
        overhead=peak_train - gpu_allocated,
        peak_infer=peak_infer,
    )

    del model, optimizer, x, output, output2, loss, loss2
    torch.cuda.empty_cache()

# ── Print table ──────────────────────────────────────────────────────────
print(f"\n{'='*95}")
print("MEMORY SUMMARY — AnomalyTransformer (batch_size=128)")
print(f"{'='*95}")
print(f"{'Dataset':<8} {'Params':>12} {'Params MB':>10} {'GPU alloc':>10} {'Peak train':>12} {'Overhead':>10} {'Peak infer':>12}")
print(f"{'-'*95}")
for name, r in results.items():
    print(f"{name:<8} {r['params']:>12,} {r['params_mb']:>9.2f} MB {r['gpu_allocated']:>9.2f} MB "
          f"{r['peak_train']:>11.2f} MB {r['overhead']:>9.2f} MB {r['peak_infer']:>11.2f} MB")
```

## Inference Speed

Measures latency, throughput, and s/iter for each dataset on CPU.
- Latency: time to process one batch (ms)
- Throughput: samples processed per second
- s/iter: latency in seconds

```python
import torch
import numpy as np
from model.AnomalyTransformer import AnomalyTransformer

device = torch.device("cuda:0")

# ── Force CUDA context initialization ─────────────────────────────────────
_ = torch.zeros(1).to(device)

DATASETS = {
    'SMD':   dict(enc_in=38, seq_len=100),
    'MSL':   dict(enc_in=55, seq_len=100),
    'SMAP':  dict(enc_in=25, seq_len=100),
    'SWaT':  dict(enc_in=51, seq_len=100),
    'PSM':   dict(enc_in=25, seq_len=100),
}

BATCH_SIZE = 128
N_RUNS     = 100   # number of timed iterations (matches ModernTCN)
N_WARMUP   = 10    # warmup runs to stabilize GPU clock and cuDNN

results = {}

for dataset_name, ds_args in DATASETS.items():
    enc_in  = ds_args['enc_in']
    seq_len = ds_args['seq_len']

    # ── Build model ────────────────────────────────────────────────────────
    model = AnomalyTransformer(
        win_size=seq_len, enc_in=enc_in, c_out=enc_in, e_layers=3
    ).float().to(device)
    model.eval()

    dummy = torch.randn(BATCH_SIZE, seq_len, enc_in, device=device)

    # ── Warmup runs ────────────────────────────────────────────────────────
    # Ensures cuDNN selects its algorithm and GPU clocks are stable
    # before any timing begins — same rationale as ModernTCN
    with torch.no_grad():
        for _ in range(N_WARMUP):
            _ = model(dummy)

    # ── Timed runs using CUDA Events ───────────────────────────────────────
    # CUDA Events measure GPU-side time directly, avoiding Python overhead
    # This matches the ModernTCN measurement methodology
    latencies = []

    with torch.no_grad():
        for _ in range(N_RUNS):
            start = torch.cuda.Event(enable_timing=True)
            end   = torch.cuda.Event(enable_timing=True)

            start.record()
            _ = model(dummy)
            end.record()

            # Synchronize to ensure GPU has finished before reading time
            torch.cuda.synchronize()
            latencies.append(start.elapsed_time(end))  # milliseconds

    latencies  = np.array(latencies)
    mean_ms    = latencies.mean()
    std_ms     = latencies.std()
    throughput = (BATCH_SIZE / (mean_ms / 1000))  # samples per second
    s_per_iter = mean_ms / 1000                   # seconds per iteration

    results[dataset_name] = {
        'seq_len':    seq_len,
        'enc_in':     enc_in,
        'mean_ms':    mean_ms,
        'std_ms':     std_ms,
        'throughput': throughput,
        's_per_iter': s_per_iter,
    }

    del model, dummy
    torch.cuda.empty_cache()

# ── Summary table ──────────────────────────────────────────────────────────
print('\n' + '=' * 98)
print(f'INFERENCE SPEED SUMMARY — AnomalyTransformer '
      f'(batch_size={BATCH_SIZE}, N={N_RUNS} runs)')
print('=' * 98)
print(f"{'Dataset':<8} {'seq_len':<9} {'enc_in':<8} {'B':<5} "
      f"{'Latency mean±std (ms)':<28} {'Throughput (samp/s)':<23} {'s/iter'}")
print('-' * 98)
for name, r in results.items():
    print(
        f"{name:<8} {r['seq_len']:<9} {r['enc_in']:<8} {BATCH_SIZE:<5} "
        f"{r['mean_ms']:.3f} ± {r['std_ms']:<19.3f}"
        f"{r['throughput']:>20,.1f}     {r['s_per_iter']:.4f}"
    )
```
