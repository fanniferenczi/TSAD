```python
import warnings
warnings.filterwarnings("ignore", message="Initializing zero-element tensors is a no-op")

# -- Count MACs for TranAD (anomaly detection) on ALL datasets ----------------
import sys
import importlib
import torch
import torch.nn as nn

# Assumes you're in TranAD/ (the notebook already does %cd there)
sys.path.insert(0, './')

# src.models imports src.constants -> src.parser.parse_args().
# In notebooks, sanitize argv to avoid argparse errors from Jupyter flags.
_argv_backup = sys.argv[:]
sys.argv = [sys.argv[0]]
from src.models import TranAD
sys.argv = _argv_backup

try:
    FlopCountAnalysis = importlib.import_module('fvcore.nn').FlopCountAnalysis
    profile = importlib.import_module('thop').profile
except ModuleNotFoundError as e:
    raise ModuleNotFoundError("Please install required packages first: pip install fvcore thop") from e

class TranADWrapper(nn.Module):
    def __init__(self, model):
        super().__init__()
        self.model = model

    def forward(self, x):
        # x: [window, batch, feats]; TranAD expects (src, tgt)
        tgt = x[-1:, :, :]
        out = self.model(x, tgt)
        return out[1] if isinstance(out, tuple) else out

# Feature dimensions per dataset (consistent with this workspace setup).
datasets = {
    'SMD':  dict(feats=38),
    'MSL':  dict(feats=55),
    'SMAP': dict(feats=25),
    'SWaT': dict(feats=51),
    'PSM':  dict(feats=25),
}

print(f"{'Dataset':<8} {'fvcore MACs':>14} {'thop MACs':>14} {'Diff%':>10}")
print('-' * 50)

for name, hp in datasets.items():
    model = TranAD(hp['feats']).double().cpu().eval()
    wrapper = TranADWrapper(model).eval()

    torch.manual_seed(42)
    dummy = torch.randn(model.n_window, 1, hp['feats'], dtype=torch.float64)

    with torch.no_grad():
        # fvcore
        fa = FlopCountAnalysis(wrapper, dummy)
        fa.unsupported_ops_warnings(False)
        fa.uncalled_modules_warnings(False)
        fvcore_macs = fa.total()

        # thop
        thop_macs, _ = profile(wrapper, inputs=(dummy,), verbose=False)

    diff_pct = abs(fvcore_macs - thop_macs) / max(fvcore_macs, 1) * 100
    print(f"{name:<8} {fvcore_macs:>14,} {thop_macs:>14,.0f} {diff_pct:>9.4f}%")
```

```python
import sys, torch, torch.nn as nn
from fvcore.nn import FlopCountAnalysis

sys.path.insert(0, './')
_argv_backup = sys.argv[:]
sys.argv = [sys.argv[0]]
from src.models import TranAD
sys.argv = _argv_backup

class TranADWrapper(nn.Module):
    def __init__(self, model):
        super().__init__()
        self.model = model
    def forward(self, x):
        elem = x[-1:, :, :]
        out = self.model(x, elem)
        return out[1] if isinstance(out, tuple) else out

def get_shape(jit_value):
    return list(jit_value.type().sizes())

def numel_from_jit(jit_value):
    shape = get_shape(jit_value)
    result = 1
    for s in shape:
        result *= s
    return result

def elementwise_handler(inputs, outputs):
    try:
        return numel_from_jit(outputs[0])
    except Exception:
        return 0

def softmax_handler(inputs, outputs):
    try:
        return 2 * numel_from_jit(outputs[0])
    except Exception:
        return 0

def addmm_handler(inputs, outputs):
    try:
        input_shape  = get_shape(inputs[1])
        weight_shape = get_shape(inputs[2])
        rows      = input_shape[0]
        inner_dim = input_shape[1]
        cols      = weight_shape[1]
        return rows * inner_dim * cols
    except Exception:
        return 0

def run_fvcore(feats):
    model = TranAD(feats).double().cpu().eval()
    wrapper = TranADWrapper(model)
    dummy = torch.randn(model.n_window, 1, feats, dtype=torch.float64)

    fa_base = FlopCountAnalysis(wrapper, dummy)
    fa_base.unsupported_ops_warnings(False)
    fa_base.uncalled_modules_warnings(False)
    base_macs = fa_base.total()
    unsupported = fa_base.unsupported_ops()

    fa_full = FlopCountAnalysis(wrapper, dummy)
    fa_full.unsupported_ops_warnings(False)
    fa_full.uncalled_modules_warnings(False)
    fa_full.set_op_handle('aten::softmax',    softmax_handler)
    fa_full.set_op_handle('aten::add',        elementwise_handler)
    fa_full.set_op_handle('aten::add_',       elementwise_handler)
    fa_full.set_op_handle('aten::mul',        elementwise_handler)
    fa_full.set_op_handle('aten::sigmoid',    elementwise_handler)
    fa_full.set_op_handle('aten::leaky_relu', elementwise_handler)
    fa_full.set_op_handle('aten::sub',        elementwise_handler)
    fa_full.set_op_handle('aten::pow',        elementwise_handler)
    fa_full.set_op_handle('aten::addmm',      addmm_handler)
    fa_full.set_op_handle('aten::repeat',     lambda i, o: 0)
    full_macs  = fa_full.total()
    by_op_full = fa_full.by_operator()

    return base_macs, full_macs, by_op_full, unsupported

datasets = {
    'SMD':   dict(feats=38),
    'MSL':   dict(feats=55),
    'SMAP':  dict(feats=25),
    'SWaT':  dict(feats=51),
    'PSM':   dict(feats=25),
}

default_ops_simplified = {'matmul', 'bmm', 'linear', 'addmm', 'mm', 'conv2d'}

for name, hp in datasets.items():
    feats = hp['feats']
    base_macs, full_macs, by_op_full, unsupported = run_fvcore(feats)

    print("=" * 65)
    print(f"  {name}  |  feats={feats}  d_model={2*feats}")
    print("=" * 65)

    print(f"\n{'Baseline fvcore (matmul/bmm only)':.<45} {base_macs:>10,}")
    print(f"{'Extended fvcore (+ elementwise + out_proj)':.<45} {full_macs:>10,}")
    print(f"{'MACs added by custom handlers':.<45} {full_macs - base_macs:>10,}")

    print("\n── Per-operator breakdown (extended) ──────────────────────────────")
    print(f"  {'Operator':<35} {'MACs':>10}  {'Counted by default?'}")
    print(f"  {'-'*35} {'-'*10}  {'-'*25}")
    for op, count in sorted(by_op_full.items(), key=lambda x: -x[1]):
        if count == 0:
            continue
        was_default = 'yes' if op in default_ops_simplified else 'no — custom handler'
        print(f"  {op:<35} {count:>10,}  {was_default}")

    print("\n── Previously unsupported, now counted ────────────────────────────")
    print(f"  {'Operator':<35} {'Occurrences':>12}  {'MACs now':>10}")
    print(f"  {'-'*35} {'-'*12}  {'-'*10}")
    for op_full_key, occ in sorted(unsupported.items(), key=lambda x: -x[1]):
        simplified = op_full_key.replace('aten::', '')
        new_count = by_op_full.get(simplified, 0)
        print(f"  {op_full_key:<35} {occ:>12}x  {new_count:>10,}")

    print("\n── Summary ─────────────────────────────────────────────────────────")
    print(f"  Baseline MACs (matmul + bmm):    {base_macs:>10,}")
    print(f"  Extended MACs:                   {full_macs:>10,}")
    print(f"  Elementwise + out_proj added:    {full_macs - base_macs:>10,}")
    pct = (full_macs - base_macs) / full_macs * 100 if full_macs > 0 else 0
    print(f"  Elementwise share of total:      {pct:>9.1f}%")
    print()
```

```python
# -- Memory usage for TranAD anomaly detection (batch_size=128) -------------------
import sys
import torch
import torch.nn as nn

# Assumes you're in TranAD/ (the notebook already does %cd there)
sys.path.insert(0, './')

# src.models imports src.constants -> src.parser.parse_args().
# In notebooks, sanitize argv to avoid argparse errors from Jupyter flags.
_argv_backup = sys.argv[:]
sys.argv = [sys.argv[0]]
from src.models import TranAD
sys.argv = _argv_backup

device = torch.device('cpu')
print(f'Using device: {device}')

# Feature dimensions per dataset (consistent with this workspace setup).
DATASETS = {
    'SMD':  dict(feats=38),
    'MSL':  dict(feats=55),
    'SMAP': dict(feats=25),
    'SWaT': dict(feats=51),
    'PSM':  dict(feats=25),
}

BATCH_SIZE = 128
results = {}

for dataset_name, ds_args in DATASETS.items():
    print(f'Measuring {dataset_name}...')
    try:
        feats = ds_args['feats']

        # Build model on CPU
        model = TranAD(feats).double().to(device)

        # Weights including buffers as actually allocated tensors on CPU
        weight_bytes = sum(p.numel() * p.element_size() for p in model.parameters())
        weight_bytes += sum(b.numel() * b.element_size() for b in model.buffers())
        weight_mem = weight_bytes / 1e6

        # TranAD input shape: [window, batch, feats]
        dummy = torch.randn(
            model.n_window, BATCH_SIZE, feats,
            device=device, dtype=torch.float64
        )
        elem = dummy[-1:, :, :]
        criterion = nn.MSELoss()

        # Train memory estimate (weights + input + output + gradients)
        model.train()
        model.zero_grad(set_to_none=True)
        out = model(dummy, elem)
        output = out[1] if isinstance(out, tuple) else out
        loss = criterion(output, dummy)
        loss.backward()

        input_mb = (dummy.numel() * dummy.element_size()) / 1e6
        output_mb = (output.numel() * output.element_size()) / 1e6
        grad_mb = sum(
            p.grad.numel() * p.grad.element_size()
            for p in model.parameters() if p.grad is not None
        ) / 1e6
        peak_train = weight_mem + input_mb + output_mb + grad_mb

        # Inference memory estimate (weights + input + output)
        model.eval()
        with torch.no_grad():
            out_eval = model(dummy, elem)
            out_infer = out_eval[1] if isinstance(out_eval, tuple) else out_eval
        infer_output_mb = (out_infer.numel() * out_infer.element_size()) / 1e6
        peak_infer = weight_mem + input_mb + infer_output_mb

        total_params = sum(p.numel() for p in model.parameters())

        results[dataset_name] = {
            'params': total_params,
            'weight_exact': total_params * 8 / 1e6,  # fp64 params x 8 bytes -> MB
            'weight_mb': weight_mem,                 # CPU allocated weights (+buffers)
            'train_mb': peak_train,
            'infer_mb': peak_infer,
            'overhead_mb': peak_train - weight_mem,
        }

        # Clean up before next dataset
        del model, dummy, elem, out, output, out_eval, out_infer, loss

    except Exception as e:
        print(f'  ERROR: {e}')
        results[dataset_name] = None

# Summary table
print('\n' + '=' * 95)
print('MEMORY SUMMARY - TranAD Anomaly Detection (batch_size=128)')
print('=' * 95)
print(f"{'Dataset':<8} {'Params':<12} {'Weights exact':<15} {'Weights CPU':<13} "
      f"{'Train peak':<13} {'Infer peak':<13} {'Overhead':<10}")
print(f"{'':>20} {'(paramsx8B)':>15} {'(allocated)':>13} "
      f"{'(MB)':>13} {'(MB)':>13} {'(MB)':>10}")
print('-' * 95)
for name, r in results.items():
    if r is None:
        print(f'{name:<8} ERROR')
    else:
        print(
            f"{name:<8} {r['params']:>10,}   {r['weight_exact']:>11.2f} MB  "
            f"{r['weight_mb']:>9.2f} MB  "
            f"{r['train_mb']:>9.2f} MB  "
            f"{r['infer_mb']:>9.2f} MB  "
            f"{r['overhead_mb']:>7.2f} MB"
        )

import sys, io
print("\n── Verification 1: independent weight size check ──────────────────")
for name, hp in DATASETS.items():
    model = TranAD(hp['feats']).double().cpu()
    
    # Method A: your code (numel × bytes per element)
    method_a = sum(p.numel() * p.element_size() 
                   for p in model.parameters()) / 1e6
    
    # Method B: serialize to buffer (includes pickle overhead → slightly larger)
    buffer = io.BytesIO()
    torch.save(model.state_dict(), buffer)
    method_b = buffer.tell() / 1e6

    # Method C: sum storage sizes directly (torch 1.8 compatible)
    seen = set()
    method_c = 0
    for p in model.parameters():
        storage_id = p.storage().data_ptr()
        if storage_id not in seen:
            seen.add(storage_id)
            # .size() returns number of elements in storage, not bytes
            method_c += p.storage().size() * p.element_size()
    method_c /= 1e6

    print(f"  {name:<8} A(numel×bytes)={method_a:.3f}MB  "
          f"B(serialized)={method_b:.3f}MB  "
          f"C(storage)={method_c:.3f}MB")
    del model

print("\n── Verification 2: overhead decomposition ─────────────────────────")
print(f"  {'Dataset':<8} {'input':>8} {'output':>8} {'grads':>8} "
      f"{'sum':>8} {'overhead':>10} {'match?'}")
print(f"  {'-'*65}")
for name, hp in DATASETS.items():
    feats = hp['feats']
    model = TranAD(feats).double().cpu().train()
    dummy = torch.randn(model.n_window, BATCH_SIZE, feats, dtype=torch.float64)
    elem  = dummy[-1:, :, :]
    
    out = model(dummy, elem)
    output = out[1] if isinstance(out, tuple) else out
    loss = nn.MSELoss()(output, dummy)
    model.zero_grad()
    loss.backward()

    input_mb  = dummy.numel()  * dummy.element_size()  / 1e6
    output_mb = output.numel() * output.element_size() / 1e6
    grad_mb   = sum(p.grad.numel() * p.grad.element_size()
                    for p in model.parameters() if p.grad is not None) / 1e6
    
    component_sum = input_mb + output_mb + grad_mb
    overhead = results[name]['overhead_mb']
    match = 'OK' if abs(component_sum - overhead) < 0.01 else 'MISMATCH'
    
    print(f"  {name:<8} {input_mb:>8.3f} {output_mb:>8.3f} {grad_mb:>8.3f} "
          f"{component_sum:>8.3f} {overhead:>10.3f} {match}")
    del model, dummy, elem, out, output, loss
```

```python
# -- Memory usage for TranAD anomaly detection (batch_size=128) -------------------
import sys
import torch
import torch.nn as nn

sys.path.insert(0, './')

_argv_backup = sys.argv[:]
sys.argv = [sys.argv[0]]
from src.models import TranAD
sys.argv = _argv_backup

device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
print(f'Using device: {device}')

DATASETS = {
    'SMD':  dict(feats=38),
    'MSL':  dict(feats=55),
    'SMAP': dict(feats=25),
    'SWaT': dict(feats=51),
    'PSM':  dict(feats=25),
}

BATCH_SIZE = 128
results = {}

def _mem_allocated():
    if device.type == 'cuda':
        torch.cuda.synchronize(device)
        return torch.cuda.memory_allocated(device) / 1e6
    return 0.0

def _peak_allocated():
    if device.type == 'cuda':
        torch.cuda.synchronize(device)
        return torch.cuda.max_memory_allocated(device) / 1e6
    return 0.0

def _reset_peak():
    if device.type == 'cuda':
        torch.cuda.reset_peak_memory_stats(device)

def _empty_cache():
    if device.type == 'cuda':
        torch.cuda.empty_cache()

for dataset_name, ds_args in DATASETS.items():
    print(f'Measuring {dataset_name}...')
    try:
        feats = ds_args['feats']

        _empty_cache()
        _reset_peak()

        model = TranAD(feats).double().to(device)
        weight_mem = _mem_allocated()

        # CPU fallback: manual weight size
        if device.type == 'cpu':
            weight_mem = (
                sum(p.numel() * p.element_size() for p in model.parameters()) +
                sum(b.numel() * b.element_size() for b in model.buffers())
            ) / 1e6

        dummy = torch.randn(model.n_window, BATCH_SIZE, feats, device=device, dtype=torch.float64)
        elem = dummy[-1:, :, :]
        criterion = nn.MSELoss()

        # Train pass — peak includes weights + input + activations + gradients
        model.train()
        model.zero_grad(set_to_none=True)
        out = model(dummy, elem)
        output = out[1] if isinstance(out, tuple) else out
        loss = criterion(output, dummy)
        loss.backward()
        peak_train = _peak_allocated()

        # CPU fallback
        if device.type == 'cpu':
            input_mb  = dummy.numel()  * dummy.element_size()  / 1e6
            output_mb = output.numel() * output.element_size() / 1e6
            grad_mb   = sum(p.grad.numel() * p.grad.element_size()
                            for p in model.parameters() if p.grad is not None) / 1e6
            peak_train = weight_mem + input_mb + output_mb + grad_mb

        # Clean up train artifacts, reset peak counter for inference
        del out, output, loss
        for p in model.parameters():
            p.grad = None
        _empty_cache()
        _reset_peak()

        # Inference pass — peak includes weights + input + output only
        model.eval()
        with torch.no_grad():
            out_eval = model(dummy, elem)
            out_infer = out_eval[1] if isinstance(out_eval, tuple) else out_eval
        peak_infer = _peak_allocated()

        if device.type == 'cpu':
            infer_output_mb = out_infer.numel() * out_infer.element_size() / 1e6
            peak_infer = weight_mem + input_mb + infer_output_mb

        total_params = sum(p.numel() for p in model.parameters())

        results[dataset_name] = {
            'params':       total_params,
            'weight_exact': total_params * 8 / 1e6,
            'weight_mb':    weight_mem,
            'train_mb':     peak_train,
            'infer_mb':     peak_infer,
            'overhead_mb':  peak_train - weight_mem,
        }

        del model, dummy, elem, out_eval, out_infer
        _empty_cache()

    except Exception as e:
        print(f'  ERROR: {e}')
        results[dataset_name] = None

# Summary table
print('\n' + '=' * 95)
print(f'MEMORY SUMMARY - TranAD Anomaly Detection (batch_size=128) [{device.type.upper()}]')
print('=' * 95)
print(f"{'Dataset':<8} {'Params':<12} {'Weights exact':<15} {'Weights alloc':<15} "
      f"{'Train peak':<13} {'Infer peak':<13} {'Overhead':<10}")
print('-' * 95)
for name, r in results.items():
    if r is None:
        print(f'{name:<8} ERROR')
    else:
        print(
            f"{name:<8} {r['params']:>10,}   {r['weight_exact']:>11.2f} MB  "
            f"{r['weight_mb']:>11.2f} MB  "
            f"{r['train_mb']:>9.2f} MB  "
            f"{r['infer_mb']:>9.2f} MB  "
            f"{r['overhead_mb']:>7.2f} MB"
        )
```

## Inference Speed

Measures latency, throughput, and s/iter for each dataset on CPU.
- Latency: time to process one batch (ms)
- Throughput: samples processed per second
- s/iter: latency in seconds

```python
# -- Inference speed for TranAD on ALL datasets (CPU timer) ------------------
import sys
import time
import numpy as np
import torch

# Assumes you're in TranAD/ (the notebook already does %cd there)
sys.path.insert(0, './')

# Reuse DATASETS/BATCH_SIZE from the Memory Usage cell above when available
if 'DATASETS' not in globals():
    DATASETS = {
        'SMD':  dict(feats=38),
        'MSL':  dict(feats=55),
        'SMAP': dict(feats=25),
        'SWaT': dict(feats=51),
        'PSM':  dict(feats=25),
    }
if 'BATCH_SIZE' not in globals():
    BATCH_SIZE = 128

# src.models imports src.constants -> src.parser.parse_args().
# In notebooks, sanitize argv to avoid argparse errors from Jupyter flags.
_argv_backup = sys.argv[:]
sys.argv = [sys.argv[0]]
from src.models import TranAD
sys.argv = _argv_backup

device = torch.device('cpu')
print(f'Using device: {device}')

N_WARMUP = 10
N_RUNS = 100

# -- Measure each dataset ------------------------------------------------------
speed_results = {}

for dataset_name, ds_args in DATASETS.items():
    print(f"Measuring {dataset_name}...", end=' ', flush=True)
    try:
        feats = int(ds_args['feats'])

        model = TranAD(feats).double().to(device).eval()
        dummy = torch.randn(model.n_window, BATCH_SIZE, feats, dtype=torch.float64, device=device)
        elem = dummy[-1:, :, :]

        # Warmup
        with torch.no_grad():
            for _ in range(N_WARMUP):
                _ = model(dummy, elem)

        # CPU timing
        times_ms = []
        with torch.no_grad():
            for _ in range(N_RUNS):
                t0 = time.perf_counter()
                _ = model(dummy, elem)
                t1 = time.perf_counter()
                times_ms.append((t1 - t0) * 1000.0)

        times_ms = np.asarray(times_ms, dtype=np.float64)

        speed_results[dataset_name] = {
            'latency_mean_ms': float(times_ms.mean()),
            'latency_std_ms': float(times_ms.std(ddof=0)),
            'latency_min_ms': float(times_ms.min()),
            'latency_max_ms': float(times_ms.max()),
            'throughput': float(BATCH_SIZE / (times_ms.mean() / 1000.0)),
            's_per_iter': float(times_ms.mean() / 1000.0),
            'window': int(model.n_window),
            'feats': feats,
            'batch_size': int(BATCH_SIZE),
        }
        print(f"done ({times_ms.mean():.2f} ms)")

        del model, dummy, elem

    except Exception as e:
        print(f"ERROR: {e}")
        speed_results[dataset_name] = None

# -- Print summary table -------------------------------------------------------
print('' + '=' * 98)
print(f"INFERENCE SPEED SUMMARY - TranAD (batch_size={BATCH_SIZE}, N={N_RUNS} runs)")
print('=' * 98)
print(f"{'Dataset':<8} {'window':<9} {'feats':<8} {'B':<4} {'Latency mean+-std (ms)':<26} {'Throughput (samp/s)':<22} {'s/iter'}")
print('-' * 98)
for name, r in speed_results.items():
    if r is None:
        print(f"{name:<8} ERROR")
    else:
        lat = f"{r['latency_mean_ms']:.3f} +- {r['latency_std_ms']:.3f}"
        print(f"{name:<8} {r['window']:<9} {r['feats']:<8} {r['batch_size']:<4} {lat:<26} {r['throughput']:>18,.1f}     {r['s_per_iter']:.4f}")
```
